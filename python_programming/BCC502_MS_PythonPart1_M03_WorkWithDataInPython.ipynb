## 1. Variables and dynamic typing

A variable is a named container that holds a value in memory.

You create one by assigning a value with the assignment operator `=`.

### Right-to-left rule

Python looks at the value on the right first, then assigns it to the name on the left.

```python
player_name = "Alex"
player_age = 25
high_score = 98.5
print(player_name, player_age, high_score)
```

You can update a variable at any time.

```python
score = 98.5
score = 100
print(score)
```

### Naming variables

- Must start with a letter or underscore, never a number (`1st_place` is illegal).
- Can only contain letters, numbers, and underscores.
- Are case-sensitive: `score` and `Score` are different containers.

By convention, Python developers use `snake_case`.

```python
student_name = "Alex"
final_score = 95
print(student_name, final_score)

score = 1
Score = 2
print(score, Score)
```

### Data types

| Type | Label | Description | Example |
|---|---|---|---|
| Integer | `int` | Whole numbers | `42`, `-7` |
| Floating point | `float` | Decimal numbers | `3.14`, `-0.5` |
| String | `str` | Text in quotes | `"Hello"`, `"42"` |

The `int` 42 and the `str` "42" are two completely different values.

### Dynamic typing

Python works out the type from the value you assign.

A variable can change type mid-program.

```python
system_status = 10
print(type(system_status))
system_status = "Active"
print(type(system_status))
```

### Checking a type

Use the built-in `type()` function.

```python
score = 98.5
print(type(score))
```

## 2. Operations with numbers

### Arithmetic operators

| Operator | Action | Example | Result | Note |
|---|---|---|---|---|
| `+` | Addition | `10 + 3` | `13` | |
| `-` | Subtraction | `10 - 3` | `7` | |
| `*` | Multiplication | `10 * 3` | `30` | |
| `/` | Division | `10 / 2` | `5.0` | Always returns a float |
| `//` | Integer division | `10 // 3` | `3` | Chops off the remainder |
| `%` | Modulo | `10 % 3` | `1` | Returns only the remainder |
| `**` | Exponentiation | `2 ** 3` | `8` | 2 to the power of 3 |

```python
print(10 + 3, 10 - 3, 10 * 3)
print(10 / 2)
print(10 // 3)
print(10 % 3)
print(2 ** 3)
```

### Understanding modulo

The `%` operator is unrelated to percentages.

It divides the first number by the second and returns what is left over.

If `number % 2` is `0`, the number is even.

Modulo can also wrap a number around: `current_item % total_items` resets to `0`.

```python
print(10 % 2, 7 % 2)

total_items = 3
for click in range(6):
    print(click, "->", click % total_items)
```

### Order of operations

Python follows the standard order (PEMDAS).

Use parentheses to force a section to run first.

```python
standard_order = 2 + 3 * 4
forced_order = (2 + 3) * 4
print(standard_order, forced_order)
```

### Integers and floats

Mixing an `int` and a `float` always returns a `float`.

```python
base_value = 10
bonus_value = 3.5
total_score = base_value + bonus_value
print(total_score, type(total_score))
```

### Updating a variable with itself

The shorthand `+=` updates a variable without repeating its name.

The same works for `-=`, `*=`, and `/=`.

```python
score = 50
score = score + 10
score += 10
print(score)
```

### Division by zero

Python does not allow it.

```python
try:
    crash_test = 10 / 0
except ZeroDivisionError as error:
    print("ZeroDivisionError:", error)
```

## 3. Convert between data types

The `input()` function always captures text.

To use it in math, convert (cast) it first.

| Function | Converts to | Example | Result |
|---|---|---|---|
| `int()` | Whole number | `int("42")` | `42` |
| `float()` | Decimal number | `float("3.14")` | `3.14` |
| `str()` | Text | `str(100)` | `"100"` |

```python
print(int("42"), float("3.14"), str(100))
```

### Converting user input

```python
raw_input = input("Enter your score: ")
score = float(raw_input)
print(score + 5)
```

### One-line shorthand

Nest `input()` inside the conversion function.

Python runs the inside function first, then passes the text to the outside function.

```python
score = float(input("Enter your score: "))
print(score)
```

### Incompatible values

If a value cannot be read as a number, Python raises a `ValueError`.

```python
try:
    user_age = int("hello")
except ValueError as error:
    print("ValueError:", error)

try:
    user_age = int("3.14")
except ValueError as error:
    print("ValueError:", error)
```

If a user might type a decimal, use `float()` instead.

Floats convert to integers, but the decimal part is always dropped.

```python
print(int(3.99))
```

### Checking a value's type

```python
score = "95"
print(type(score))
score = float(score)
print(type(score))
```

## Practice: calculate a student's GPA

Ask for three grades, convert them to numbers, and print the average to two decimal places.

Hint: `f"{value:.2f}"` formats a number with two decimals.

```python
# TODO 1: ask for three grades with input() and convert each with float()
# TODO 2: add them up and divide by 3
# TODO 3: print the GPA with two decimal places
```

### Solution

Saved as `p1m3_gpa_calculator.py`.

```python
def calculate_gpa(grades):
    """Return the average of a list of grade points."""
    return sum(grades) / len(grades)


def main():
    grades = []
    for course in ("first", "second", "third"):
        grades.append(float(input(f"Enter the grade points for the {course} course: ")))
    gpa = calculate_gpa(grades)
    print(f"Your GPA is {gpa:.2f}")


main()
```

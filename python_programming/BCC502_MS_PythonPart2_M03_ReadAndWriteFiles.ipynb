## 1. Open files safely with context managers

Opening a file gives you a file object to read from or write to.

When you finish, close the file so the operating system can release resources and save pending writes.

### The problem with forgetting to close

If an error happens after `open()` but before `.close()`, the file stays open.

That can cause lost data, locked files, or resource leaks.

```python
file = open("notes.txt", "r")
contents = file.read()
print(contents)
file.close()
```

### Using the with statement

The `with` statement is a context manager.

It closes the file automatically when the block finishes, even if an error occurs.

Everything after `as` is the name of the file object inside the block.

```python
with open("notes.txt", "r") as file:
    contents = file.read()
    print(contents)
# The file is automatically closed here
print("Closed:", file.closed)
```

### File modes

| Mode | Meaning |
|---|---|
| `"r"` | Read (default). The file must already exist. |
| `"w"` | Write. Creates a new file or overwrites the existing one. |
| `"a"` | Append. Adds to the end, or creates the file if missing. |
| `"r+"` | Read and write. |

Using the wrong mode is a common cause of accidentally erased files.

For the rest of this notebook, always open files with `with`.

## 2. Read data from files

### Read the entire file at once

`.read()` returns the whole file as one string.

It works well for small files, but loads everything into memory.

```python
with open("notes.txt", "r") as file:
    contents = file.read()
print(contents)
```

If the file does not exist, read mode raises a `FileNotFoundError`.

```python
try:
    with open("missing.txt", "r") as file:
        print(file.read())
except FileNotFoundError as error:
    print("FileNotFoundError:", error)
```

### Read the file as a list of lines

`.readlines()` returns a list of strings.

Each string keeps its newline character `\n`, so use `.strip()` to remove it.

```python
with open("notes.txt", "r") as file:
    lines = file.readlines()
print(lines)
for line in lines:
    print(line.strip())
```

### Read one line at a time

Iterating over the file object is the most memory-friendly way.

Only one line is held in memory at a time.

```python
with open("notes.txt", "r") as file:
    for line in file:
        print(line.strip())
```

### Choosing the right approach

| If you want to... | Use... |
|---|---|
| Work with the entire file as one string | `.read()` |
| Get a list of lines you can index into | `.readlines()` |
| Process each line efficiently | `for line in file` |

## 3. Write data to files

### Writing with "w"

Write mode creates a new file or completely overwrites an existing one.

`.write()` does not add a newline, unlike `print()`, so add `\n` yourself.

> **Warning:** write mode erases the file the moment it is opened.

```python
with open("greeting.txt", "w") as file:
    file.write("Hello, world!\n")
    file.write("Welcome to Python.\n")

with open("greeting.txt") as file:
    print(file.read())
```

### Appending with "a"

Append mode adds to the end and keeps what is already there.

If the file does not exist, Python creates it.

It is a good fit for log files and notes.

```python
with open("log.txt", "a") as file:
    file.write("New entry added.\n")

with open("log.txt") as file:
    print(file.read())
```

### Writing multiple lines

`.writelines()` writes a list of strings.

It does not add newlines, so include them yourself.

```python
tasks = ["Buy groceries\n", "Walk the dog\n", "Finish homework\n"]
with open("tasks.txt", "w") as file:
    file.writelines(tasks)

with open("tasks.txt") as file:
    print(file.read())
```

### Choosing between write and append

| Use "w" when... | Use "a" when... |
|---|---|
| You want to start fresh each time | You want to keep the existing content |
| You save the final result of a process | You record new events or entries |
| Overwriting is intentional | Preserving history matters |

## Practice: build a to-do list

Build a to-do list app that saves tasks to a file so they persist between runs.

Load tasks when the program starts, add new ones, and save them when you quit.

```python
# TODO 1: read tasks.txt if it exists (handle FileNotFoundError)
# TODO 2: ask for new tasks until the user types "quit"
# TODO 3: write all tasks back to tasks.txt
```

### Solution

Saved as `p2m3_todo_list.py`.

Run this cell twice to see tasks persist between runs.

```python
FILENAME = "tasks.txt"


def load_tasks(filename=FILENAME):
    """Read tasks from the file. Return an empty list if it does not exist yet."""
    try:
        with open(filename, "r") as file:
            return [line.strip() for line in file if line.strip()]
    except FileNotFoundError:
        return []


def save_tasks(tasks, filename=FILENAME):
    """Write all tasks to the file, one per line."""
    with open(filename, "w") as file:
        for task in tasks:
            file.write(task + "\n")


def main():
    tasks = load_tasks()
    print(f"You have {len(tasks)} task(s).")
    for number, task in enumerate(tasks, start=1):
        print(f"{number}. {task}")
    while True:
        new_task = input("New task (or type 'quit'): ").strip()
        if new_task.lower() == "quit":
            break
        tasks.append(new_task)
    save_tasks(tasks)
    print("Tasks saved.")


main()
```

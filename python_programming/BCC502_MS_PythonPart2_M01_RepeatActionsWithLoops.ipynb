## 1. Choose between for and while loops

A loop repeats a block of code automatically.

Python has two loop types: `for` and `while`.

Choose by whether you know how many times to repeat ahead of time.

### The for loop with range()

Use a `for` loop for a known number of repetitions, or to go through each item in a collection.

`range(5)` produces 0 through 4.

It starts at 0 and stops before the number you give.

```python
for i in range(5):
    print(i)
```

### range(start, stop, step)

The loop stops before it reaches the stop value.

```python
for number in range(1, 10, 2):    # start at 1, stop before 10, step by 2
    print(number)
```

### The while loop

Use a `while` loop to repeat as long as a condition stays `True`.

Modify the condition inside the loop so it eventually stops.

```python
health = 3
while health > 0:
    print(f"Player is alive. Health: {health}")
    health -= 1
```

### Avoid infinite loops

A `while` loop checks its condition before every iteration.

If the condition is always `True`, the loop never stops and freezes your program.

Always update the condition so it eventually becomes `False`.

### Choosing the right loop

**Use a `for` loop when**

- You know how many times to repeat.
- You want an automatic exit after a set number of repetitions.
- You want a counter variable.

**Use a `while` loop when**

- The number of repetitions depends on a variable state.
- You are waiting for something, such as user input.
- You need control over when the loop stops.

## 2. Controlling loops with break and continue

### break: exit the loop early

`break` ends the loop immediately.

Python moves on to the next line after the loop.

```python
for count in range(10):
    if count == 5:
        break
    print(count)
print("Loop ended.")
```

### continue: skip to the next iteration

`continue` skips the rest of the loop body and jumps to the next iteration.

```python
for count in range(6):
    if count == 3:
        continue
    print(count)
```

### break in a while loop

`while True` with `break` is a common pattern for unpredictable user input.

Type `exit` to leave the loop.

```python
while True:
    user_input = input("Type 'exit' to stop the loop: ")
    if user_input.lower() == "exit":
        break
    print(f"You entered: {user_input}")
print("Goodbye!")
```

### continue in a while loop

Update the condition before `continue`, or you can create an infinite loop.

```python
current_slot = 0
while current_slot < 5:
    current_slot += 1
    if current_slot == 3:
        print(f"Skipping item slot {current_slot}.")
        continue
    print(f"Successfully processed item in slot {current_slot}.")

print("Inventory scan complete!")
```

### Using else with loops

A loop's `else` block runs only when the loop finishes without hitting `break`.

```python
color = "blue"
for i in range(3):
    user_input = input("Guess my favorite color: ")
    if user_input.lower() == color:
        print("You guessed it!")
        break
else:
    print(f"Sorry, the correct answer is {color}.")
```

## Practice: guess the number

Pick a random number from 1 to 100 with `random.randint`.

Use a `while` loop to keep asking until the player guesses it.

Print "Too low." or "Too high." after each wrong guess.

```python
import random

# TODO 1: pick a secret number
# TODO 2: loop until the guess is correct
# TODO 3: use break to leave the loop
```

### Solution

Saved as `p2m1_guess_the_number.py`.

```python
import random


def main():
    secret = random.randint(1, 100)
    attempts = 0
    while True:
        guess = int(input("Guess a number between 1 and 100: "))
        attempts += 1
        if guess < secret:
            print("Too low.")
        elif guess > secret:
            print("Too high.")
        else:
            print(f"You got it in {attempts} attempts!")
            break


main()
```

## 1. Store and manage data with lists

A list is an ordered collection written inside square brackets, with items separated by commas.

Lists can hold any type and can be changed after creation.

```python
colors = ["red", "green", "blue"]
scores = [95, 82, 76, 91]
print(colors)
print(scores)
```

### Accessing items by index

Each item has a numbered position called an index, starting at `0`.

Negative indexes count from the end.

```python
colors = ["red", "green", "blue"]
print(colors[0])
print(colors[2])
print(colors[-1])
print(colors[-2])
```

### Common list operations

| Operation | What it does |
|---|---|
| `len(list)` | Number of items |
| `list.append(value)` | Adds to the end |
| `list.remove(value)` | Removes the first match |
| `list[i] = value` | Overwrites index `i` |
| `list.pop()` | Removes and returns the last item |
| `list.sort()` | Sorts in place |

```python
colors = ["red", "green", "blue"]
print(len(colors))
colors.append("yellow")
print(colors)
colors.remove("red")
print(colors)
colors[0] = "purple"
print(colors)
```

```python
numbers = [3, 1, 2]
last = numbers.pop()
numbers.sort()
print(last, numbers)
```

### Iterating over a list

Lists work naturally with `for` loops.

```python
scores = [95, 82, 76, 91]
for score in scores:
    print(score)
```

## 2. Store and organize data with dictionaries

A dictionary stores key-value pairs.

You use a meaningful label, called a key, instead of a position.

Dictionaries use curly braces, with colons between keys and values.

```python
user = {
    "name": "Alex",
    "age": 25,
    "email": "alex@example.com"
}
print(user)
```

### Accessing values by key

```python
print(user["name"])
print(user["age"])
```

A missing key raises a `KeyError`.

The `.get()` method returns `None`, or a default you choose.

```python
try:
    print(user["phone"])
except KeyError as error:
    print("KeyError:", error)

print(user.get("phone"))
print(user.get("phone", "not listed"))
```

### Common dictionary operations

| Operation | What it does |
|---|---|
| `dict[key] = value` | Adds or updates a key |
| `del dict[key]` | Removes a key-value pair |
| `key in dict` | `True` if the key exists |
| `dict.keys()` | All the keys |
| `dict.values()` | All the values |
| `dict.items()` | Key-value pairs |

```python
user = {"name": "Alex", "age": 25}
user["email"] = "alex@example.com"
user["age"] = 26
del user["age"]
print("name" in user)
print(user)
print(list(user.keys()))
print(list(user.values()))
```

### Iterating over a dictionary

The `.items()` method gives you both keys and values.

```python
user = {"name": "Alex", "age": 25, "email": "alex@example.com"}
for key, value in user.items():
    print(f"{key}: {value}")
```

### Lists vs. dictionaries

| Use a list when... | Use a dictionary when... |
|---|---|
| Order matters | Each value has a meaningful label |
| You access items by position | You access values by key |
| You have a sequence of similar items | You have a set of named properties |

## Practice: customer order

Build a coffee shop order tracker.

Store the menu in a dictionary, the order in a list, and print a formatted receipt with a loop.

```python
menu = {"latte": 4.50, "espresso": 3.00, "muffin": 3.25, "cookie": 2.00}

# TODO 1: build an order list with a few menu items
# TODO 2: loop over the order and look up each price in menu
# TODO 3: print each line and the total
```

### Solution

Saved as `p2m2_order_tracker.py`.

```python
MENU = {"latte": 4.50, "espresso": 3.00, "muffin": 3.25, "cookie": 2.00}


def receipt_lines(order, menu=MENU):
    """Return the receipt as a list of text lines."""
    lines = []
    total = 0
    for item in order:
        price = menu[item]
        lines.append(f"{item.title():<12} ${price:>6.2f}")
        total += price
    lines.append(f"{'Total':<12} ${total:>6.2f}")
    return lines


def main():
    order = []
    while True:
        item = input("Add an item (or type 'done'): ").strip().lower()
        if item == "done":
            break
        if item in MENU:
            order.append(item)
        else:
            print(f"Sorry, we do not have {item}.")
    print("\n".join(receipt_lines(order)))


main()
```

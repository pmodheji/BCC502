## 1. Create functions with parameters

A function is a named, reusable block of code that runs when you call it.

Define one with `def`, a name, parentheses, and a colon.

Function names should be lowercase with underscores.

```python
def greet():
    print("Hello, world!")

greet()
```

### Adding parameters

Parameters are named inputs inside the parentheses.

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Alex")
greet("Sam")
```

Separate multiple parameters with commas.

```python
def greet(name, greeting):
    print(f"{greeting}, {name}!")

greet("Alex", "Hi")
```

### Default parameter values

Use `=` in the definition to give a parameter a default.

If the caller leaves it out, the default is used.

```python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Alex")
greet("Alex", "Welcome")
```

### Keyword arguments

Pass arguments by name to make code easier to read.

They can appear in any order, after any positional arguments.

```python
greet(name="Alex", greeting="Welcome")
greet(greeting="Hi", name="Sam")
```

## 2. Return values from functions

Use `return` to send a value back to the caller.

Once `return` runs, the function ends immediately.

```python
def add(a, b):
    return a + b

result = add(3, 4)
print(result)
print(add(10, 5) * 2)
```

### Return vs. print

- `print()` displays a value on the console.
- `return` gives a value back so it can be stored, used, or passed on.

A function that only prints returns `None`.

```python
def add_and_print(a, b):
    print(a + b)

x = add_and_print(3, 4)
print(x)
```

### Returning multiple values

Separate the values with commas.

Python packs them into a tuple that you can unpack.

```python
def get_coordinates():
    return 10.0, 20.0

x, y = get_coordinates()
print(f"x: {x}, y: {y}")
```

### Functions with no return value

A function without `return` still returns `None`.

That is fine when its job is to do something, like print or write to a file.

```python
def greet(name):
    print(f"Hello, {name}!")

result = greet("Alex")
print(result)
```

## 3. Manage variable scope

Scope is the region of your code where a variable exists.

### Local variables

A variable created inside a function is local.

It exists only while the function runs.

```python
def calculate_total(price, tax):
    total = price + tax       # total is local to this function
    return total

calculate_total(10, 2)
try:
    print(total)
except NameError as error:
    print("NameError:", error)
```

### Global variables

A variable created outside any function is global.

Functions can read globals, but by default they cannot change them.

```python
message = "Hello"

def greet():
    print(message)

greet()
```

Assigning to a name that also exists globally creates a new local variable.

The global is left alone.

```python
color = "blue"

def change_color():
    color = "red"        # a new local variable

change_color()
print(color)
```

### The global keyword (use sparingly)

Declare a name `global` to change it from inside a function.

A better pattern is to pass the value in and return the updated value.

```python
count = 0

def increment():
    global count
    count = count + 1

increment()
print(count)
```

```python
def increment(count):
    return count + 1

count = 0
count = increment(count)
print(count)
```

### Why scope matters

- Local variables keep functions independent.
- Reading globals is fine for constants such as configuration values.
- Writing to globals from inside functions is a common source of hard-to-find bugs.

Prefer parameters and return values.

## 4. Create importable modules

A module is just a `.py` file whose functions and variables can be reused from other files.

The module name is the file name without `.py`.

This cell creates `math_utils.py` in the notebook's folder.

```python
%%writefile math_utils.py
# math_utils.py
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

if __name__ == "__main__":
    # Quick self-test, only runs when this file is executed directly
    print(add(2, 3))
```

### Importing a module

```python
import math_utils

print(math_utils.add(2, 3))
print(math_utils.multiply(4, 5))
```

### Importing specific names

You can then use them without the module prefix.

```python
from math_utils import add, multiply

print(add(2, 3))
print(multiply(4, 5))
```

### Using an alias

Give a long module name a shorter alias with `as`.

You often see this with libraries, for example `import pandas as pd`.

```python
import math_utils as mu

print(mu.add(2, 3))
```

### The `if __name__ == "__main__":` block

Importing a file runs its code.

Put code that should run only when the file is executed directly inside this block.

Importing `math_utils` above printed nothing extra.

Running it directly runs the self-test.

```python
import subprocess, sys

result = subprocess.run([sys.executable, "math_utils.py"], capture_output=True, text=True)
print("Run directly:", result.stdout.strip())
```

### Why modules matter

- They keep each file focused on one responsibility.
- They make code easier to test.
- They make it natural to share and reuse code across projects.

Build a calculator that uses functions for its operations.

Handle division by zero without crashing.

### Solution

The operations live in `calc_operations.py`, and the main program is `p2m4_calculator.py`.

This cell writes the module and then imports it.

```python
%%writefile calc_operations.py
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return None
    return a / b
```

```python
from calc_operations import add, subtract, multiply, divide

OPERATIONS = {"+": add, "-": subtract, "*": multiply, "/": divide}


def calculate(a, operator, b):
    """Apply an operator to two numbers. Return None if it is not valid."""
    operation = OPERATIONS.get(operator)
    if operation is None:
        return None
    return operation(a, b)


def main():
    a = float(input("First number: "))
    operator = input("Operator (+, -, *, /): ").strip()
    b = float(input("Second number: "))
    result = calculate(a, operator, b)
    if result is None:
        print("That calculation is not valid.")
    else:
        print(f"{a} {operator} {b} = {result}")


main()
```

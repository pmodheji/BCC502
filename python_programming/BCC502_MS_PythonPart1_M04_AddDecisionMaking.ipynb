## 1. Booleans and comparison operators

A Boolean (`bool`) has exactly two values: `True` or `False`.

They are the building blocks of all decision-making.

```python
is_game_over = True
is_user_logged_in = False
print(type(is_game_over))
```

Python is case-sensitive.

Writing `true` or `false` raises a `NameError`.

```python
try:
    value = true
except NameError as error:
    print("NameError:", error)
```

### Comparison operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `10 > 3` | `True` |
| `<` | Less than | `3 < 10` | `True` |
| `>=` | Greater than or equal to | `5 >= 5` | `True` |
| `<=` | Less than or equal to | `4 <= 3` | `False` |

```python
score = 85
print(score >= 90)
print(score >= 70)
```

### Comparing strings

```python
entered_password = "secret123"
correct_password = "secret123"
print(entered_password == correct_password)
```

### Equality and assignment

- Single equals `=` assigns a value.
- Double equals `==` compares two values.

Using `=` inside a comparison causes an error.

```python
score = 100
try:
    print(score = 50)
except TypeError as error:
    print("TypeError:", error)
```

## 2. Conditional logic: if, elif, else

An `if` statement runs a block only when its condition is `True`.

The block must be indented.

### Colon and indentation rules

- Put a colon `:` at the end of the conditional line.
- Use the same indentation (four spaces) for every line in the block.
- Unindent the next line to end the block.

```python
score = 85
if score >= 90:
    print("Congratulations!")
    print("Grade: A")
print("Program complete.")
```

### Adding an alternative with else

```python
score = 75
if score >= 90:
    print("Grade: A")
else:
    print("Grade: B or lower")
```

### Chaining with elif

Python checks each condition in order.

The first `True` condition runs and the rest are skipped.

If none is `True`, the `else` block runs.

```python
score = 85
if score >= 90:
    print("Grade: A")
elif score >= 80:     # Python stops here: 85 >= 80 is True
    print("Grade: B")
elif score >= 70:     # skipped, even though 85 >= 70
    print("Grade: C")
else:
    print("Grade: D or lower")
```

Order your conditions from most specific to least specific.

## 3. Logical operators: and, or, not

| Operator | What it does | Example | Result |
|---|---|---|---|
| `and` | `True` only if both sides are `True` | `True and False` | `False` |
| `or` | `True` if at least one side is `True` | `True or False` | `True` |
| `not` | Flips a Boolean | `not True` | `False` |

```python
print(True and False)
print(True or False)
print(not True)
```

### Using and

```python
age = 20
has_ticket = True
if age >= 18 and has_ticket:
    print("Entry granted.")
else:
    print("Entry denied.")
```

### Using or

In everyday language, "or" often means one or the other, but not both.

In Python, `or` means one or the other, or both.

```python
is_weekend = False
is_holiday = True
if is_weekend or is_holiday:
    print("The office is closed.")
```

### Using not

```python
is_logged_in = False
if not is_logged_in:
    print("Please log in to continue.")
```

### Combining operators

Always use parentheses to group conditions.

```python
score = 85
has_extra_credit = True
is_past_due = False
if (score >= 90 or has_extra_credit) and not is_past_due:
    print("Requirement met.")
```

### Avoid shortcut errors

`score == 90 or 100` is always `True`, because `100` is a nonzero number.

Compare each value explicitly.

```python
score = 50
if score == 90 or 100:
    print("Incorrect: this prints for any score!")

if score == 90 or score == 100:
    print("Correct version")
else:
    print("Correct version: score is not 90 or 100")
```

## Practice: rock, paper, scissors

Write a function `decide(player, computer)` that returns `"win"`, `"lose"`, or `"tie"`.

Then use `random.choice` to pick the computer's move.

Rock beats scissors, scissors beats paper, and paper beats rock.

```python
# TODO 1: define decide(player, computer) using if, elif, else
# TODO 2: pick the computer's choice with random.choice
# TODO 3: ask the player and print the result
```

### Solution

Saved as `p1m4_rock_paper_scissors.py`.

```python
import random

CHOICES = ("rock", "paper", "scissors")


def decide(player, computer):
    """Return 'win', 'lose', or 'tie' from the player's point of view."""
    if player == computer:
        return "tie"
    elif (player == "rock" and computer == "scissors") or \
         (player == "paper" and computer == "rock") or \
         (player == "scissors" and computer == "paper"):
        return "win"
    else:
        return "lose"


def main():
    player = input("Choose rock, paper, or scissors: ").strip().lower()
    if player not in CHOICES:
        print("That is not a valid choice.")
        return
    computer = random.choice(CHOICES)
    print(f"Computer chose {computer}.")
    result = decide(player, computer)
    if result == "tie":
        print("It's a tie!")
    elif result == "win":
        print("You win!")
    else:
        print("You lose!")


main()
```

## 1. Install Anaconda and VS Code

To write and run Python on your own machine, you need Python and a code editor.

### Install Anaconda

Anaconda Distribution is a free Python distribution. It installs Python, the conda package and environment manager, and many popular packages in one step.

1. Go to anaconda.com/download.
2. Download the installer for your operating system.
3. Run the installer and accept the defaults. On Windows, use the **Anaconda Prompt** from the Start menu.

Confirm the installation by running these commands in the Anaconda Prompt (Windows) or Terminal (macOS and Linux):

```
conda --version
python --version
```

You should see version numbers such as `conda 24.11.3` and `Python 3.12.10`.

This cell checks the Python that is running this notebook.

````python
import sys
print(sys.version)
print("Executable:", sys.executable)
````

### Install Visual Studio Code

VS Code is a free, lightweight code editor from Microsoft for Windows, macOS, and Linux.

1. Go to code.visualstudio.com.
2. Download and run the installer for your operating system.
3. Open VS Code after the installation completes.

### Install the Python extension

1. Open the Extensions panel with Ctrl+Shift+X.
2. Search for **Python** and select the extension published by Microsoft.
3. Click **Install**.

VS Code can then detect the Python that comes with Anaconda.

## 2. Environments and packages

A package is a reusable code library that extends what Python can do.

A conda environment is an isolated virtual environment with its own copy of Python and its own packages.

Packages in one environment do not affect another.

### Create and activate a conda environment

In the Anaconda Prompt or Terminal, run this command. It creates an environment named `myproject` that uses Python 3.12. Type `y` when conda asks to proceed.

```
conda create --name myproject python=3.12
```

| Task | Command |
|---|---|
| Activate | `conda activate myproject` |
| Deactivate | `conda deactivate` |
| List all environments | `conda env list` |

The commands are the same on Windows, macOS, and Linux.

When an environment is active, `(myproject)` appears at the start of your terminal prompt.

This cell shows which environment is running this notebook.

````python
import os, sys
print("Conda environment:", os.environ.get("CONDA_DEFAULT_ENV", "none (not started from conda)"))
print("Environment folder:", sys.prefix)
````

### Install packages with conda

conda is the package manager that comes with Anaconda.

| Task | Command |
|---|---|
| Install a package | `conda install requests` |
| List installed packages | `conda list` |
| Save the environment to a file | `conda env export > environment.yml` |
| Recreate it from the file | `conda env create -f environment.yml` |

VS Code detects your conda environments automatically.

Look for the Python version in the bottom status bar, or run **Python: Select Interpreter**, to choose your `myproject` environment.

This cell lists a few installed packages, like `conda list` does.

````python
import importlib.metadata as metadata

names = sorted(d.metadata["Name"] for d in metadata.distributions() if d.metadata["Name"])
print(len(names), "packages installed")
for name in names[:10]:
    print(name, metadata.version(name))
````

## 3. Create and run a Python application

1. In VS Code, open a folder for your project (File > Open Folder).
2. In the Explorer panel, click the New File icon and name it `app.py`.
3. Add the code below.

The `.py` extension tells VS Code this is a Python file.

````python
name = input("What is your name? ")
print(f"Hello, {name}! Welcome to VS Code.")
````

### Run your application

**Option 1, Run button:** click the Run Python File button (the play icon) in the top-right corner of the editor.

**Option 2, Terminal:** with your conda environment active, run `python app.py`.

The cell below does what Option 2 does: it saves `app.py` and runs it as a separate Python process.

````python
import subprocess, sys, tempfile, os

folder = tempfile.mkdtemp()
path = os.path.join(folder, "app.py")
with open(path, "w") as file:
    file.write('name = "Alex"\nprint(f"Hello, {name}! Welcome to VS Code.")\n')

result = subprocess.run([sys.executable, path], capture_output=True, text=True)
print(result.stdout)
````

### Read error messages

Error messages tell you what went wrong and on which line.

Reading them carefully is one of the most important debugging skills.

This cell runs a program with a typo in a variable name.

````python
import subprocess, sys, tempfile, os

folder = tempfile.mkdtemp()
path = os.path.join(folder, "app.py")
with open(path, "w") as file:
    file.write('message = "Hi"\nprint(mesage)\n')

result = subprocess.run([sys.executable, path], capture_output=True, text=True)
print(result.stderr)
````

The traceback shows a `NameError` on line 2.

It is a typo: `mesage` should be `message`.

## 4. Write code with GitHub Copilot

GitHub Copilot is an AI coding assistant built into VS Code.

It watches your code and offers suggestions, completes lines, fills in functions, and catches common mistakes.

### Sign up and install

1. Create a GitHub account at github.com and select **Copilot Free**.
2. In VS Code, open Extensions (Ctrl+Shift+X).
3. Search for **GitHub Copilot Chat** and install the extension published by GitHub.
4. Sign in with your GitHub account.

### How Copilot helps as you type

Copilot shows inline suggestions in gray text.

Press **Tab** to accept one, or keep typing to ignore it.

For example, after you type `print("How old are you?")`, Copilot might suggest `age = input()`.

### Quick fixes

When VS Code highlights an error, a light bulb icon often appears.

Click it to see **Fix using Copilot**.

### Guide Copilot with comments

Write a comment such as `# Prompt the user for their name and greet them`.

Copilot suggests code that matches your comment on the next line.

### Chatting with Copilot

Open the chat view from the chat icon in the title bar.

Shortcut: Ctrl+Alt+I on Windows and Linux, Control+Command+I on Mac.

### What Copilot is, and is not

Its suggestions can be incorrect, incomplete, or wrong for your situation.

> **Important:** always read and understand what Copilot suggests before accepting it. You are still the programmer.

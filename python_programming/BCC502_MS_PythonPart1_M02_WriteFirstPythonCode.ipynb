## 1. Display output with print

The most fundamental thing a program can do is communicate with you.

In Python, you display output with the built-in `print()` function.

Pass the text inside parentheses, enclosed in quotes.

```python
print("Hello, world!")
```

### Single and double quotes

Python does not care which you use, as long as you match them up.

```python
print("Hello, world!")
print('Hello, world!')
```

### Avoid missing quotes

Without quotes, Python looks for a variable instead of text.

The `#` symbol starts a comment, a note for humans that Python ignores.

```python
# This code is missing quotes around the text
try:
    print(Hello)
except NameError as error:
    print("NameError:", error)
```

### Printing multiple items

Separate values with commas.

Python inserts a space between each item.

```python
print("Hello", "world!")
```

### Escape characters

Escape characters start with a backslash and act as secret commands inside your text.

| Sequence | What it does |
|---|---|
| `\n` | Moves text to a new line |
| `\t` | Adds a tab space |
| `\\` | Prints a literal backslash |
| `\"` | Prints a double quote |

```python
print("Line 1\nLine 2")
print("Name:\tAlex")
print("C:\\Users")
print("She said, \"Hi!\"")
```

Use the backslash and not the forward slash.

Typing `/n` just prints `/n`.

## 2. Accept user input

The `input()` function pauses your program and waits for the user to type and press Enter.

To save the answer, catch it with a variable.

```python
favorite_color = input("What is your favorite color? ")
print("Oh, I love", favorite_color, "too!")
```

Add a space or new line at the end of the prompt.

Without it, the typing is glued to your text, like `What is your favorite color?Blue`.

### A quick word about data types

Every value has a data type, a label that tells Python what kind of value it is.

- `str` (string): text in quotes, like `"Blue"` or `"25"`.
- `int` (integer): a whole number without quotes, like `25`.

You can add numbers, but you cannot add a number to text.

```python
print(type("25"))
print(type(25))
print(25 + 1)
```

### Input is always text

The `input()` function always returns text, even if the user types a number.

```python
age = input("How old are you? ")
print(type(age))
try:
    print("Next year you will be", age + 1)
except TypeError as error:
    print("TypeError:", error)
```

### How to read this error

`TypeError` means you are mixing two data types that do not belong together.

"Concatenate" means to glue text together.

Python sees `"25" + 1` and does not know how to glue a number to text.

## 3. Manipulate strings

A string is a sequence of characters in single or double quotes.

### Concatenation

Join strings with the `+` operator.

```python
first = "Hello"
last = "world"
message = first + ", " + last + "!"
print(message)
```

### A better way: f-strings

Place the letter `f` directly before the opening quote.

Put variable names inside curly braces.

```python
name = "Alex"
print(f"Hello, {name}!")
```

If you forget the `f`, Python prints the braces literally.

```python
name = "Alex"
print("Hello, {name}!")
```

### f-strings and data types

Math inside the braces still needs a number, so convert the input with `int()` first.

Note: the Microsoft Learn unit shows `{age + 1}` here, but that still raises a `TypeError` when `age` is text.

```python
age = input("How old are you? ")
print(f"You are {age} years old.")
print(f"Next year you will be {int(age) + 1}")
```

### Common string methods

| Method | What it does | Example | Result |
|---|---|---|---|
| `.upper()` | Uppercase | `"hello".upper()` | `"HELLO"` |
| `.lower()` | Lowercase | `"HELLO".lower()` | `"hello"` |
| `.strip()` | Trims blank spaces from the ends | `" hi ".strip()` | `"hi"` |
| `.replace(old, new)` | Swaps characters | `"cat".replace("c", "b")` | `"bat"` |

```python
print("hello".upper())
print("HELLO".lower())
print(" hi ".strip())
print("cat".replace("c", "b"))
```

### Counting characters with len()

Use the standalone `len()` function.

```python
username = "Alex!"
print(len(username))
```

### String methods return a new string

Methods do not change your original variable.

To keep the change, save the result.

```python
greeting = "hello"
greeting.upper()
print(greeting)          # still lowercase

greeting = greeting.upper()
print(greeting)          # now uppercase
```

## Practice: create a personalized greeting

Write a program that asks for the user's name and displays a personalized greeting.

Use `input()`, `.strip()`, `.title()`, and an f-string.

Fill in the cell below, then compare with the solution.

```python
# TODO 1: ask the user for their name with input()
# TODO 2: clean it up with .strip() and .title()
# TODO 3: print a greeting with an f-string
```

### Solution

The same program is saved as `p1m2_personalized_greeting.py`.

```python
def build_greeting(raw_name):
    """Clean up a name and return a personalized greeting."""
    name = raw_name.strip().title()
    return f"Hello, {name}! Welcome to Python."


def main():
    raw_name = input("What is your name? ")
    print(build_greeting(raw_name))
    print(f"Your name has {len(raw_name.strip())} characters.")


main()
```
